```python
class Student:
    scname="OM AL KORA"
    def __init__(self,name,age,score):
        self.__name=name
        self.age=age
        self.score=score

    def getname(self):
        return self.__name

    def setname(self,name):
        result=self.__name=name
        return result 

    def describeobject(self):
        return f"my name is {self.__name} and my age is {self.age}  and I get {self.score}"
    @classmethod
    def describclass(cls):
        cls.result=f"this is class method"
        return cls.result
    
```
