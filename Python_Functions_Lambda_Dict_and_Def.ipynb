```python
def max3(a,b,c):
    if a>b and a>c:
        return f"{a} is greater than {b} and {c}"
    elif b>a and b>c:
        return f"{b} is greater than {a} and {c}"
    else:
        return f"{c} is greater than {a}, {b}"
```

```python
max3(1,5,-1)
```
