## 5. Abstraction


```python
class Animal:
    Zoo_name="Giza Zoo"
    def __init__(self,name,type):
        self.__name=name
        self.__type=type
    def getname(self):
        return self.__name
    def setname(self,new_name):
        self.__name=new_name
    def gettype(self):
        return self.__type
    def settype(self,newtype):
        self.__type=newtype
    def describe(self):
        return f"this is animal called {self.__name} and it is {self.__type}"
```

```python
lion=Animal("lion","Cats")
#print(lion.name)


print(lion.getname())
print(lion.gettype())
print(lion.describe())

lion.setname("Tiger")
lion.settype("People")

print(lion.getname())
print(lion.gettype())
print(lion.describe())
```

## 6. Inheritance


```python
class Person:
    def __init__(self,name,age,scname):
        self.__name=name
        self.__age=age
        self.__scname=scname

    def getname(self):
        return self.__name

    def setname(self,newname):
        self.__name=newname
    
    def getage(self):
        return self.__age

    def setage(self,newage):
        self.__age=newage
    

    def getscname(self):
        return self.__scname

    def setscname(self,newscname):
        self.scname=newscname

    def describe(self):
        return f"my name is {self.__name} and my age is {self.__age} and my school name is {self.__scname}"

    
  

```

```python
class Student(Person):
    def __init__(self,name,age,scname,courses):
        super().__init__(name,age,scname)
        self.__courses =courses

    def getcourses(self):
        return self.__courses

    def setcourses(self,newcourses):
        self.__courses=newcourses

    def describe(self):
        super().describe()
        print("and I took courses like "+ self.__courses)
    
```

```python
std1=Student("Ali",20,"Elsharabia","Math")
print(std1.getname())
print(std1.getcourses())
std2=Person("Ali",20,"Elsharabia")

print(std1.describe())
```
