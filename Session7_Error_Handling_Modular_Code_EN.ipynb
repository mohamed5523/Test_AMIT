### Handling multiple exceptions
You can either use multiple `except` blocks or a tuple of exception types.

### `else` and `finally`
- **`else`**: runs only if no exception occurred in `try` block.  
- **`finally`**: always runs (useful for cleanup, closing resources).

# 🏦 Question: Simple Banking System (Using OOP and Import in Python)

## Problem Statement  
Design and implement a simple **Banking System** using **Object-Oriented Programming (OOP)** in Python.  
Your solution must follow a **modular design**:  

- Create a file named **`models.py`** that contains the `BankAccount` class.  
- In a separate file (e.g., **`main.py`** or a Jupyter Notebook), **import the class** from `models.py` and demonstrate the system.  

---

## Requirements  

### Create a `BankAccount` class  
- Attributes:  
  - `account_number` (integer)  
  - `owner` (string)  
  - `balance` (float, default `0.0`)  

- Methods:  
  - `deposit(amount)` → increases balance by the given amount (only if positive).  
  - `withdraw(amount)` → decreases balance if sufficient funds exist, otherwise shows an error.  
  - `__str__()` → returns a string showing account number, owner, and current balance.  

---

## In the main program (or notebook):  
1. Import `BankAccount` from `models.py`.  
2. Create at least **two bank accounts**.  
3. Display their details.  
4. Deposit money into one account.  
5. Withdraw money from the other account.  
6. Display updated balances.
